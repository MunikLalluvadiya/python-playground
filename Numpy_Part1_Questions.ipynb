Q6.
Demonstrate the 'slices are views, not copies' behavior: create an array, slice it, modify the slice, and print both the original and the slice to show the original changed too. Then fix it using .copy().


```python
arr = np.array([10, 20, 30, 40, 50])
sl = arr[1:4]        # this is a VIEW, not a copy

sl[0] = 999            # modify the slice
print("Original after modifying slice:", arr)
print("Slice:", sl)

# Fix using .copy()
arr2 = np.array([10, 20, 30, 40, 50])
sl_copy = arr2[1:4].copy()
sl_copy[0] = 999
print("\nOriginal after modifying the copy:", arr2)
print("Copy:", sl_copy)
```

Q8.
Create an array with mixed integers and floats (e.g., [1, 2, 3.5]), print its dtype, and explain in a comment why NumPy chose that type.


```python
arr = np.array([1, 2, 3.5])
print(arr)
print("Dtype:", arr.dtype)

# NumPy arrays must hold a SINGLE dtype for all elements (unlike Python lists).
# Since 3.5 can't be represented as an int without losing data, NumPy
# "upcasts" the entire array to float64 so no information is lost —
# the ints 1 and 2 become 1.0 and 2.0.
```

Q9.
Create a float array [3.2, 7.8, 1.1, 9.9] and convert it to an integer array using .astype(int). Print both arrays and explain what happened to the decimal values.


```python
floats = np.array([3.2, 7.8, 1.1, 9.9])
ints = floats.astype(int)

print("Float array:", floats)
print("Integer array:", ints)

# .astype(int) TRUNCATES (chops off) the decimal part — it does NOT round.
# 3.2 -> 3, 7.8 -> 7 (not 8!), 1.1 -> 1, 9.9 -> 9 (not 10!)
```

Q10.
Create a 3x4 array of zeros using np.zeros(), then use indexing to set the entire second row to the value 5 and the entire third column to the value 8. Print the final array.


```python
arr = np.zeros((3, 4))
print("Original:\n", arr)

arr[1, :] = 5   # entire second row (index 1)
arr[:, 2] = 8   # entire third column (index 2)

print("\nFinal array:\n", arr)
```
