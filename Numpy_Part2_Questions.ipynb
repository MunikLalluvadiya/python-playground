Given an array of integers, find the largest and smallest element without using min()/max().

```python
def find_min_max(arr):
    largest = arr[0]
    smallest = arr[0]
    
    for num in arr:
        if num >= largest:
            largest = num
        elif num <= smallest: 
            smallest = num
    return int(smallest), int(largest)

arr = np.array([15,10,48,35,62])
print(find_min_max(arr))
```

Reverse an array in-place without using slicing ([::-1]) or a built-in reverse method.

```python
arr = np.array([1,2,3,4,5])

start = 0
end = -1
mid = arr.size//2

while start<mid:
    temp = arr[start]
    arr[start] = arr[end]
    arr[end] = temp
    start += 1
    end -= 1
print(arr)
```

Find the sum and average of all elements in an array using a loop.

```python
arr = np.array([1,2,3,4])

sum = 0
for i in arr:
    sum += int(i)

avr = sum / arr.size

print(f" Sum : {sum} \n Avrage : {avr}")
```

Count how many even and odd numbers are in an array.

```python
arr = np.array([1,2,3,4,5])
odd = 0
even = 0
for i in arr:
    if int(i)%2 == 0:
        even+=1
    else:
        odd+=1
print(f" odd:{odd} \n even:{even}")
```

Given an array, remove all duplicate elements while preserving the original order.

```python
arr = np.array([1,2,3,5,4,2,1])
Seen = []
result = []

for i in arr:
    if int(i) not in Seen:
        result.append(int(i))
        Seen.append(int(i))
print(result)

    




```

Find the second largest element in an array in a single pass (no sorting allowed).

```python
arr = np.array([-1,-2,-3,-5,-4])
large = min(arr)
sec_large = min(arr)

for i in arr:
    if i>large:
        sec_large = large
        large = i
    elif sec_large<i<large:
        sec_large = i

print(sec_large)
```
