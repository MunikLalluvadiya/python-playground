Create two 1D arrays of 5 numbers each, and print their sum, difference, and element-wise product.



```python
import numpy as np

arr1 = np.array([10,20,30,40,50])
arr2 = np.array([15,25,35,45,55])

Sum_Cal = arr1 + arr2
print(Sum_Cal)

Diffrence = arr2 - arr1
print(Diffrence)

product = arr1 * arr2
print(product)

```

Q6.
Create an array of 15 numbers. Use boolean indexing to extract only the numbers that are both greater than 10 AND even.


```python
arr = np.arange(15)

Result = arr[(arr > 10) & (arr % 2 == 0)]
print(Result)
```
