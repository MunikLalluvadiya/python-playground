Q2. Write a program with a Rectangle class having length and width, with methods area() and perimeter().

```python
class rectangle:

    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        Calculate_area = self.length * self.width
        print(f"Area of Rectangle Is :{Calculate_area}")

    def perimeter(self):
        Calculate_perimeter = 2*(self.length + self.width)
        print(f"perimeter of Rectangle Is :{Calculate_perimeter}")

Rac1 = rectangle(1,5)
Rac1.area()
Rac1.perimeter()
```

Q9.
Create a class Temperature with attribute celsius, and instance methods to_fahrenheit() and to_kelvin() that return the converted values (F = C * 9/5 + 32, K = C + 273.15).


```python
class Temperature:
    
    def __init__(self,celsius):
        self.celsius = celsius

    def to_fahrenheit(self):
        f = self.celsius * 9/5 + 32
        print(f"{self.celsius} celsius = {f} fahrenheit")

    def to_kelvin(self):
        k = self.celsius + 273.15
        print(f"{self.celsius} celsius = {k} kelvin ")

Temp_1 = Temperature(500)
Temp_1.to_fahrenheit()
Temp_1.to_kelvin()


    
```
