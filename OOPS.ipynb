Q1. Write a program that creates a Car class with attributes brand, model, and year. Create 2 objects and print their details.

Q2. Write a program that creates a Student class with a method display_info() that prints the student's name and marks.

```python
class Student :
    
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks

    def display_info(self):
        print(f"Student Name : {self.name} \n Student Marks : {self.marks}")
        

Student1 = Student("Hiren",85)
Student2 = Student("Rahul",95)
Student1.display_info()
Student2.display_info()
```

Q5. Write a program with a Rectangle class having length and width, with methods area() and perimeter().

```python
class rectangle:

    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        Calculate_area = self.length * self.width
        print(f"Area of Rectangle Is :{Calculate_area}")

    def perimeter(self):
        Calculate_perimeter = 2*(self.length + self.width)
        print(f"perimeter of Rectangle Is :{Calculate_perimeter}")

Rac1 = rectangle(1,5)
Rac1.area()
Rac1.perimeter()
```

Q8. Write a program with a BankAccount class where balance is a private attribute , accessible only through deposit() and withdraw() methods.

```python
class BankAccount:
    def __init__(self,ACC_NO,Bal):
        self.Acc = ACC_NO
        self.__bal = Bal

    def deposit(self,Amount):
        self.__bal += Amount
        print(f"Available Balance: {self.__bal}")

    def Withdraw(self,Amount):
        self.__bal -= Amount
        print(f"Available Balance: {self.__bal}")


User1 = BankAccount(123456,20000)
User1.deposit(10000)
User1.Withdraw(10000)

    
```

Q10. Write a program with a base class Animal (with a speak() method) and a derived class Dog that overrides speak().

```python
class Animal:
    def speak(self,name):
        print(f"{self.name} Make Sound.")

class dog(Animal):
    def __init__(self,name):
        self.name = name
        
Dog = dog("Tomy")
Dog.speak("Tomy")
```
