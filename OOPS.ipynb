Q2. Write a program with a Rectangle class having length and width, with methods area() and perimeter().

```python
class rectangle:

    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        Calculate_area = self.length * self.width
        print(f"Area of Rectangle Is :{Calculate_area}")

    def perimeter(self):
        Calculate_perimeter = 2*(self.length + self.width)
        print(f"perimeter of Rectangle Is :{Calculate_perimeter}")

Rac1 = rectangle(1,5)
Rac1.area()
Rac1.perimeter()
```

Q4.
Create a class Counter with an instance attribute count starting at 0, and instance methods increment() (adds 1) and reset() (sets back to 0). Create an object, call increment() three times, then print the count.

```python
class Counter:
    def __init__(self):
        self.count = 0

    def increment(self):
        self.count += 1

    def reset(self):
        self.count -= 1

Counter_1 = Counter()
Counter_1.increment()
Counter_1.increment()
Counter_1.increment()
print(Counter_1.count)

    
```

Q6.
Create a class Circle with attribute radius and instance methods calculate_area() and calculate_circumference() (use 3.14159 for pi). Create an object with radius 7 and print both results.


```python
class Circle:
    pi = 3.14159
    def __init__(self,radius):
        self.radius = radius

    def Area(self):
        Area = self.pi * (self.radius**2)
        return Area
        

    def Circumference(self):
        Circumference = 2*(self.pi * self.radius)
        print(Circumference)

Cir_1 = Circle(7)
print(Cir_1.Area())
(Cir_1.Circumference())


```

Q8.
Create a class Library that stores a list of Book objects (from Q5) internally, with a method add_book() to add a new book and a method list_books() to print all book titles.


```python
class Library:
    def __init__(self):
        self.books = []   

    def add_book(self, book):
        self.books.append(book)

    def list_books(self):
        for book in self.books:
            print(book.title)


# Testing it
library = Library()

book1 = Book("The Alchemist", "Paulo Coelho", 399)
book2 = Book("Atomic Habits", "James Clear", 499)

library.add_book(book1)
library.add_book(book2)

library.list_books()
```

Q9.
Create a class Temperature with attribute celsius, and instance methods to_fahrenheit() and to_kelvin() that return the converted values (F = C * 9/5 + 32, K = C + 273.15).


```python
class Temperature:
    
    def __init__(self,celsius):
        self.celsius = celsius

    def to_fahrenheit(self):
        f = self.celsius * 9/5 + 32
        print(f"{self.celsius} celsius = {f} fahrenheit")

    def to_kelvin(self):
        k = self.celsius + 273.15
        print(f"{self.celsius} celsius = {k} kelvin ")

Temp_1 = Temperature(15)
Temp_1.to_fahrenheit()
Temp_1.to_kelvin()


    
```

Q10.
Create a class Student with name and a list of marks. Add a method highest_mark() that returns the highest mark, and a method is_passing() that returns True if the average is 40 or above.


```python
class Student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks

    def highest_mark(self):
        return max(self.marks)

    def is_passing(self):
        avr = sum(self.marks)/len(self.marks)
        return avr >= 40 
        

Std1 = Student("Narendr",[65, 23, 65])
print(Std1.name)
print(Std1.highest_mark())        
print(Std1.is_passing())        
```

Q11. Write a program with a BankAccount class where balance is a private attribute , accessible only through deposit() and withdraw() methods.

```python
class BankAccount:
    def __init__(self,ACC_NO,Bal):
        self.Acc = ACC_NO
        self.__bal = Bal

    def deposit(self,Amount):
        self.__bal += Amount
        print(f"Available Balance: {self.__bal}")

    def Withdraw(self,Amount):
        self.__bal -= Amount
        print(f"Available Balance: {self.__bal}")


User1 = BankAccount(123456,20000)
User1.deposit(10000)
User1.Withdraw(10000)

    
```
