Given a DataFrame of employees with an age column, filter it to show only employees where age >= 25 and age < 40, using & . 

```python
Employees = {
    "Name" : ["Jethlal","Abdul","Popatlal"],
    "Age" : [56,35,39]
}

df = pd.DataFrame(Employees)
print(df[(df["Age"] >= 25) & (df["Age"] < 40 )])
```

Add a new column bonus to a DataFrame, calculated as 10% of an existing salary column. Then modify the salary column itself by giving everyone a flat raise of 5000.

```python
Employees = {
    "Name" : ["Jethlal","Abdul","Popatlal"],
    "Salary" : [50000,12000,15000]
}

df = pd.DataFrame(Employees)

df["Bonus"] = df["Salary"] * 0.10

df["Salary"] += 5000

print(df)
```
