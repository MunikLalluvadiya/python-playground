Q2.
Create a DataFrame from a dictionary with columns: product, price, and quantity (at least 5 rows). Print df.head(), df.shape, and df.info().


```python
data = {
    "Product" : [ "TV","Watch","Phone","Fridge","Laptop"],
    "Price" : [20000,2500,1500,10000,22000],
    "Quantity" : [10,20,10,20,10]
}
df = pd.DataFrame(data,index=[1,2,3,4,5])
print(df)
```

Q3.

```python
print(df.loc[:,["Product","Price"]])
```

Q4.

Q6.
Using the DataFrame from Q2, add a new column called total_value that equals price multiplied by quantity


```python
df["total_value"] = df["Price"]*df["Quantity"]
print(df)
```

Q7.
Using the DataFrame from Q2, filter and print only the rows where price is greater than a value of your choice


```python
new = df[df["Price"] > 10000]
print(new)
```

Q8.
Using the DataFrame from Q2, filter rows where price is greater than X AND quantity is greater than Y (your choice of values), using proper combined boolean syntax.


```python
new = df[(df["Price"] > 5000 ) & (df["Quantity"]> 10)]
print(new)
```
