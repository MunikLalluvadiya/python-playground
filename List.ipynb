Q1.
Write a program that creates a list of 5 favorite movies, then uses append(), insert(), and remove() to modify it, printing the list after each operation

```python
movies = ["Inception", "Interstellar", "The Matrix", "Titanic", "Iron Man"]
print("Original list:", movies)

movies.append("Avatar")
print("After append('Avatar'):", movies)

movies.insert(2, "The Dark Knight")
print("After insert(2, 'The Dark Knight'):", movies)

movies.remove("Titanic")
print("After remove('Titanic'):", movies)
```

Q2.
Write a program that takes 5 numbers as input from the user, stores them in a list, and prints the maximum, minimum, and sum without using max()/min()/sum() — using a loop instead.


```python
numbers = list(map(int, input("Enter 5 numbers separated by spaces: ").split()))

maximum = 0
minimum = 2**99
total = 0

for num in numbers:
    total += num

    if num >= maximum:
        maximum = num
    if num <= minimum:
        minimum = num

print(f"Maximum is: {maximum}, Minimum is: {minimum}, Sum is: {total}")
```

Q3.
Write a program using list comprehension to create a list of squares of all numbers from 1 to 20 that are divisible by 3.


```python
squares = [num ** 2 for num in range(1, 21) if num % 3 == 0]

print("Squares of numbers from 1 to 20 divisible by 3:", squares)
```

Q4.
Write a program that removes duplicate values from a list while preserving the original order (do not use the set() data type — use a loop and an empty list).


```python
original_list = [1, 2, 2, 3, 4, 4, 4, 5, 1, 6]
result = []

for item in original_list:
    if item not in result:
        result.append(item)

print("Original list:", original_list)
print("List without duplicates:", result)
```

Q5.
Write a program that creates a 3x3 nested list (matrix) and calculates the sum of all elements using nested loops.


```python
matrix = [
    [12, 7, 33],
    [4, 25, 9],
    [18, 2, 41]
]

total = 0

for row in matrix:
    for num in row:
        total += num

print("Sum of all elements:", total)
```

Q6.
Write a program to demonstrate the shallow copy pitfall: create a list, assign it to a second variable WITHOUT using .copy(), modify the second variable, and show that the original also changes. Then fix it using .copy().


```python

original_list = [10,20,30,40,50]
refrence_list =  original_list

print(f"original_list is : {original_list}")
print(f"refrence_list is : {refrence_list}")

refrence_list[0] = 99

print("____After Modification____")
print(f"original list is : {original_list}")
print(f"refrence list is : {refrence_list}")

copy_list = refrence_list.copy()
copy_list[0] = 10

print(f"copy list is : {copy_list}")
print(f"original_list is still : {original_list}")
```

Q9.
Write a program using sort() with a lambda key to sort a list of tuples [('Item A', 250), ('Item B', 90), ('Item C', 400)] by price in ascending order.


```python
items = [('Item A', 250), ('Item B', 90), ('Item C', 400)]

items.sort(key=lambda x: x[1])

print(f"Sorted by price : {items}")
```

Q10.
Write a program using tuple unpacking with * to split a list of 7 numbers into the first number, the last number, and a list of everything in between.
