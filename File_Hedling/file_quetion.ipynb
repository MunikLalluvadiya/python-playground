Q5.
Write a program that creates a CSV file named products.csv with columns name, price, quantity, and writes 4 sample product rows into it using the csv module.


```python
import csv
with open("products.csv","w",newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Name","Price","Quantity "])
    writer.writerow(["Leptop",25000,30])
    writer.writerow(["Watch",2800,40 ])
    writer.writerow(["TV",70000,10])
    writer.writerow(["Mobile",30000,10 ])
```

Q6.
Write a program that reads the products.csv file created in Q5 using csv.DictReader, and prints the total value (price * quantity) of all products combined.


```python
import csv
total = 0
with open("products.csv","r") as file:
    dictionary = csv.DictReader(file)

    for row in dictionary:
        price= float(row["Price"])
        quantity= float(row["Quantity "])
        total += price * quantity
print(total)
    
```

Write a program using the math module to calculate the area and circumference of a circle, given its radius as input (use math.pi).


```python
from math import pi

Redius = int(input("Enter redius :"))
Area = pi * (Redius**2)
print(f" Area Of Circle is :{Area}")
circumference = 2* pi * Redius
print(f" circumference Of Circle is :{circumference}")
```

Q10.
Write a combined program: generate 5 random numbers between 1 and 100 using the random module, write them to a file called random_numbers.txt (one per line), then read the file back and print their sum and average using math functions where relevant.


```python
import random

Numbers = []
for i in range(5):
    num = random.randint(1,100)
    Numbers.append(num)


with open("random_number.txt","w") as file:
    for j in range(5):
        file.write(str(Numbers[j]))
        file.write("\n")


File_Numbers = []
with open("random_number.txt","r") as file:
    for row in file:
        File_Numbers.append(int(row))


from math import fsum

Sum = fsum(File_Numbers)
Avr = Sum/len(File_Numbers)

print(f"List From File :{File_Numbers}")
print(f"Sum :{Sum}")
print(f"Avrage :{Avr}")

```
