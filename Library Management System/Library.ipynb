```python
import csv
def Add_Book():
    with open("BOOKS.CSV","a",newline="") as file:
        writer = csv.writer(file)

        book_id = int(input("Enter Book ID :"))
        title = input("Enter Book Titel :")
        author = input("Enter Author Name :")
        quantity = int(input("Enter Total Quantity :"))
        status = int(input("Enter Available Copys :"))

        writer.writerow([book_id, title, author, quantity, status])
        print("Book Added Sucsessfully..")


def Remove_Book():
    with open("BOOKS.CSV","r") as file:
        reader = csv.reader(file)
        book_id = (input("Enter Book ID :"))
        Writer_list = []

        is_print = True
        for Book in reader:
            if Book[0] == book_id :
                print("Remove Sucsessfull..")
                is_print = False
            else:
                Writer_list.append(Book)
        
        if is_print :
            print("Book Not Available..")

        with open("BOOKS.CSV","w",newline="") as file_w :
            writer = csv.writer(file_w)

            for row in Writer_list:
                writer.writerow(row)



def Search_Book():
    with open("BOOKS.CSV","r") as file:
            reader = csv.reader(file)
            book_id = (input("Enter Book ID :"))
            
            is_print = True            
            for deta in reader:
                if deta[0] == book_id:
                    is_print = False
                    print(deta)
            if is_print :
                print("BOOK NOT FOUNd.")


def Add_Member():
    with open("MEMBER.CSV","a",newline="") as file:
            writer = csv.writer(file)

            member_id = int(input("Enter Member ID :"))
            member_name = input("Enter Your Name :")
            borrow_book = int(input("How Many Book You Borrow :"))

            writer.writerow([member_id, member_name, borrow_book])
            print("Member Added Sucsessfully..")


def Remove_Member():
    with open("MEMBER.CSV","r") as file:
        reader = csv.reader(file)
        Members_List = []
        Member_id = (input("Enter Member Id :"))

        is_print = True
        for row in reader:
            if row[0] == Member_id:
                is_print = False
                print("Remove Sucsess.")
            else:
                Members_List.append(row)

        if is_print:
            print("Member Not Found.")
       

        with open("MEMBER.CSV","w",newline="") as file_w:
            writer = csv.writer(file_w)

            for deta in Members_List:
                writer.writerow(deta)

        
def Search_Member():
    with open("MEMBER.CSV","r") as file:
        reader = csv.reader(file)
        Member_id = (input("Enter Member Id :"))

        is_print = True
        for row in reader:
            if row[0] == Member_id:
                is_print = False
                print(row)

        if is_print:
                print("Member Not Found.")


while True:
    print("_________MENU________")
    print(" 1-Add Book\n 2-Remove Book\n 3-Search Book\n 4-Add Member\n 5-Remove Member\n 6-Search Member\n 0-Exit")
    choice = int(input("Enter Your Choice :"))
    if choice == 1:
        Add_Book()
    elif choice == 2:
        Remove_Book()
    elif choice == 3:
        Search_Book()
    elif choice == 4:
        Add_Member()
    elif choice == 5:
        Remove_Member()
    elif choice == 6:
        Search_Member()
    elif choice == 0:
        break
    else:
        print("Invalid Input.")

                
    



   
```
