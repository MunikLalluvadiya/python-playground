Q2.
Write a program that takes a sentence as input and creates a dictionary counting how many times each word appears in it.


```python
Sentence = input("Enter Santence: ").split()
Word_Count = {}

for word in Sentence:
    Count = Sentence.count(word)
    Word_Count[word]=Count

print(Word_Count)
```

Q4.
Write a program that stores a nested dictionary of 3 employees (each with name, department, and salary), then loops through and prints a formatted summary for each employee using f-strings.


```python
Employees = {
    "Employee_1":{"Name":"Rohit Sharma","Department":"HR","Salary":80000},
    "Employee_2":{"Name":"Virat Kohli","Department":"IT","Salary":79000}

}

for Emp,Details in Employees.items():
    print(Emp)
    for Key,Value in Details.items():
        print(Key,":",Value)

```

Q9.
Write a program that takes a dictionary of student marks and uses items() combined with a loop to find and print the name of the student with the HIGHEST marks.


```python
Marks = {
    87:"Rahul",
    91:"Sohil",
    76:"Koushik",
    60:"Vishal"
}
Highest = 0
for mrk in Marks:
    if mrk > Highest:
        Highest = mrk
print(f"{Marks[Highest]} Is Pass With {Highest} Marks")
```
